# SIPaKMeD Federated Learning and Centralized Baseline

This notebook records the 15 federated CPU runs (3 algorithms × 3, 4, 5, 10, and 20 clients) plus a **centralized one-client FedAvg baseline**. Each used 10 rounds × 2 local epochs, batch size 8, image size 96, seed 42, and the same stratified split (2,620 train / 656 validation images; 1,638 per binary label). In the one-client run, the sole client holds the entire training split; this is centralized training in this single-process simulation. Federated runs partition that same training split among clients.

**These are 10-round experimental results, not necessarily converged or thesis-final results.** They are from a single seed and CPU simulation.   python .\federated_sipakmed.py --data .\data\sipakmed_binary --clients $clients --algorithm fedprox --mu 0.01 --rounds 10 --local-epochs 2 --batch-size 8 --image-size 96 --workers 0 --seed 42
  python .\federated_sipakmed.py --data .\data\sipakmed_binary --clients $clients --algorithm clustered --clusters 3 --personalization-lambda 0.01 --rounds 10 --local-epochs 2 --batch-size 8 --image-size 96 --workers 0 --seed 42
}
